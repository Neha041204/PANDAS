```python
df["Bonus"] = 5
```

```python
df["FinalMarks"] = df["Marks"] + df["Bonus"]
```

```python
print(df)
```

```python
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)
```

```python
print(df)
```

```python
def grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    else:
        return "D"
df["Grade"] = df["Marks"].apply(grade)
print(df)
```

```python
print(df.sort_values("Marks", ascending=False))
```

```python
print(df.nlargest(3, "Marks"))
```

```python
df["Rank"] = df["Marks"].rank(ascending=False)
print(df)
```

```python
print(df.groupby("Department")["Marks"].count())
```

```python
print(df.groupby("Department")["Marks"].max())
```

```python
print(df.groupby("Department")["Marks"].min())
```

```python
print(df.groupby("Department")["Marks"].agg(
    ["count", "mean", "max", "min"]
))
```

```python
merit_list = df[["Rank", "RollNo", "Name", "Department", "Marks", "Grade", "Result"]]
```

```python
merit_list = merit_list.sort_values("Marks", ascending=False)
```

```python
merit_list = merit_list.reset_index(drop=True)
```
